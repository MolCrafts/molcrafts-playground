# molpy

This kernel is Pyodide. `molpy` and its dependencies (`molrs`, `mollog`, `molcfg`, `numpy`) are already installed. `numpy` comes from the Pyodide distribution, not from a PyPI manylinux wheel.


```python
import molpy as mp

ir = mp.io.smiles.SmilesIr("CCO")
print("ethanol IR components:", ir.n_components)

mol = mp.io.smiles.SmilesIr("c1ccccc1").to_atomistic()
print("benzene atoms:", mol.n_atoms)
```
