# molvis

`molvis` is installed. Structure coercion and palette rendering run in this kernel.

`mv.Stage()` opens a viewer by hosting a page over a local WebSocket. Pyodide has no socket to bind, so that path does not run here. The same page, driven in-process, is the MolVis app at [app.molcrafts.org/molvis/](https://app.molcrafts.org/molvis/).


```python
import molpy as mp
import molvis as mv

mol = mp.io.smiles.SmilesIr("CCO").to_atomistic()
frame = mv.coerce_to_frame(mol)
payload, buffers = mv.frame_payload(mol)
print(type(frame).__name__, "atoms", mol.n_atoms)
print("wire buffers", len(buffers))
```

```python
from IPython.display import Image

from molvis.palettes import PaletteEntry, render_palette_preview

png = render_palette_preview([
    PaletteEntry("C", "#404040"),
    PaletteEntry("O", "#E04040"),
    PaletteEntry("N", "#3050F8"),
    PaletteEntry("H", "#FFFFFF"),
])
Image(data=png)
```
